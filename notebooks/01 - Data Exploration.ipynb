   ### Data Visualisation ###

```python
# Compter le nombre de produits par classe
counts_per_class = Y_train['prdtypecode'].value_counts()

# Créer une palette de couleurs catégorielles 
palette = sns.color_palette("husl", len(counts_per_class))

plt.figure(figsize=(16, 8))
counts_per_class.plot(kind='bar', color=palette)
plt.title('Nombre de produits par classe')
plt.xlabel('Classe')
plt.ylabel('Nombres de produits')
plt.show()
```

   ### Image ##

```python
# Fusionner le contenu des colonnes "imageid" et "productid" dans une nouvelle colonne 'nom_img' en ajoutant des texte
X_train['nom_img'] = 'image_' + X_train['imageid'].astype(str) + '_product_' + X_train['productid'].astype(str)
X_train.head()
## Cela va permettre d'importer les images
```

```python
## Fusion des dataFrame:
df_train = X_train.join(Y_train)

display(df_train.head())
```

```python
for category in prdtypecode_values:
    nb_images = 5  # Nombre d'images à afficher
    # Récupère le nb_images à afficher pour la catégorie en cours
    images = df_train[df_train['prdtypecode'] == category]['nom_img'].head(nb_images)

    fig, axes = plt.subplots(1, nb_images, figsize=(30, 8))  # Taille plus grande de la figure

    for i, image_name in enumerate(images):
        image = plt.imread(r"../resources/images/image_train/" + image_name + ".jpg")
        axes[i].imshow(image)
        axes[i].set_title(f"Catégorie {category} - Image {i+1}") 
        axes[i].axis('off')  
        axes[i].set_aspect('auto')  

    plt.tight_layout()  # Ajuster automatiquement les sous-tracés pour un meilleur ajustement
    plt.show()
```
