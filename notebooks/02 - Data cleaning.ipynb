```python
import pandas as pd
import nltk
import html
from nltk.corpus import stopwords
import re
from unidecode import unidecode
```

```python
# Création du Dataframe X_test
X_test=pd.read_csv('../resources/X_test_update.csv', index_col=0)
X_test.head()
```

```python
# Création du Dataframe X_train
X_train=pd.read_csv('../resources/X_train_update.csv', index_col=0)
X_train.head()
```

```python
# Création du Dataframe df_train
df_train=pd.read_csv('../resources/generated/df_train.csv', index_col=0)
display(df_train.head())
```

```python
# Remplacer les valeurs NaN par une chaîne vide
X_train['description'] = X_train['description'].fillna('')
X_test['description'] = X_test['description'].fillna('')

print("Valeurs nulles restantes =", X_train.isna().sum().sum() + X_test.isna().sum().sum())
```

```python
# Fusionner le contenu des colonnes "designation" et "description" dans une nouvelle colonne 'Texte'
X_train['Texte'] = X_train['designation'] + ' ' + X_train['description']
X_test['Texte'] = X_test['designation'] + ' ' + X_test['description']
```

```python
# Accéder au contenu de la colonne 'Texte' de la ligne spécifiée afin de vérifier si la fusion a été faite:
contenu_texte = X_test['Texte'].iloc[3]

# Afficher le contenu complet de la colonne 'Texte' de la ligne spécifiée
print('\n',contenu_texte,'\n')
```

```python
# Remplacement des entités html échappées, ex: "d&#39;un" --> "d'un"
X_train['Texte'] = X_train['Texte'].apply(html.unescape)
X_test['Texte'] = X_test['Texte'].apply(html.unescape)
# Vérification que les entités HTMl ont été remplacées :
print(X_test['Texte'].iloc[3])
```

```python
# Fonction pour tout mettre en miniscule
def lower_case(Texte):
    Texte = Texte.lower().strip()
    return Texte

X_train['Texte'] = X_train['Texte'].apply(lower_case)
X_test['Texte'] = X_test['Texte'].apply(lower_case)
# Vérification que le texte a été mis en minuscule
print(X_test['Texte'].iloc[3])
```

```python
# Remplacement des accents, ç, etc.
X_train['Texte'] = X_train['Texte'].apply(unidecode)
X_test['Texte'] = X_test['Texte'].apply(unidecode)
# Vérification du remplacement :
print(X_test['Texte'].iloc[3])
```

```python
# supprimer les tags HTML
def remove_htmltags(Texte):
    Texte = re.sub('<[^<]+?>', ' ', Texte)
    return Texte

X_train['Texte'] = X_train['Texte'].apply(remove_htmltags)
X_test['Texte'] = X_test['Texte'].apply(remove_htmltags)

# Vérification du remplacement :
print(X_test['Texte'].iloc[3])
```

```python
# remplacer les caractère qui ne sont pas des lettres par un espace
def keeping_alphabetical_only(Texte):
    Texte = re.sub(r"[^a-z]+", " ", Texte)
    return Texte

X_train['Texte'] = X_train['Texte'].apply(keeping_alphabetical_only)
X_test['Texte'] = X_test['Texte'].apply(keeping_alphabetical_only)
# Vérification du remplacement :
print(X_test['Texte'].iloc[3])
```

```python
# transformer la colonne 'Texte' en string
display(X_train['Texte'].dtypes)
X_train['Texte'] = X_train['Texte'].astype('string')
X_test['Texte'] = X_test['Texte'].astype('string')
display(X_train['Texte'].dtypes)
```

```python
# Séparer les mots de texte en série des mots
X_train['Texte'] = X_train['Texte'].str.split()
X_test['Texte'] = X_test['Texte'].str.split()
```

```python
# fonction pour garder les mots unique de texte
def unique_texte(Texte):
    mots_unique = []
    for word in Texte:
        if word not in mots_unique:
            mots_unique.append(word)
    return mots_unique
```

```python
# Appliquer la fonction mots_unque
X_train['Texte'] = X_train['Texte'].apply(unique_texte)
X_test['Texte'] = X_test['Texte'].apply(unique_texte)
```

```python
# Import des packages -- à faire que la première fois
nltk.download('punkt')
nltk.download('stopwords')
```

```python
# Initialiser la variable des mots vides
stop_words = stopwords.words('french') + stopwords.words('english') + stopwords.words('german') +['plus', 'peut', 'tout', 'etre', 'sans', 'dont', 'aussi', 'comme', 'meme',
                                                                                                  'bien','leurs', 'elles', 'cette','celui', 'ainsi', 'encore', 'alors', 'toujours',
                                                                                                  'toute','deux', 'nouveau', 'peu', 'car', 'autre', 'jusqu', 'quand', 'ici', 'ceux', 
                                                                                                  'enfin', 'jamais', 'autant', 'tant', 'avoir', 'moin', 'celle', 'tous', 'contre',
                                                                                                  'pourtant', 'quelque', 'toutes', 'surtout', 'cet', 'comment', 'rien', 'avant',
                                                                                                  'doit', 'autre', 'depuis', 'moins', 'tre', 'souvent', 'etait', 'pouvoir', 'apre', 
                                                                                                  'non', 'ver', 'quel', 'pourquoi', 'certain', 'fait', 'faire', 'sou', 'donc', 'trop',
                                                                                                  'quelques', 'parfois', 'tres', 'donc', 'dire', 'entre', 'sous', 'quelle']
```

```python
# filtrer les mots qui ne sont pas des stopwords et avec une longueur sup ou égale à 2
def filter_stop_words(list):
    result_list=[]
    for mot in list:
        if (mot not in stop_words and len(mot)>=2):
            result_list.append(mot)
    return result_list
```

```python
# Applique le filtre et vérifie le résultat
X_train['Texte']= X_train['Texte'].apply(filter_stop_words)
X_test['Texte']= X_test['Texte'].apply(filter_stop_words)
X_test['Texte'].iloc[3]
```

```python
# joindre les séries des mots
X_train['Texte'] = X_train['Texte'].apply(lambda x: " ".join(x))
X_test['Texte'] = X_test['Texte'].apply(lambda x: " ".join(x))
X_test['Texte'].iloc[3]
```
