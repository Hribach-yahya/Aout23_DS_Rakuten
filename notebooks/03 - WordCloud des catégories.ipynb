```python
import pandas as pd
from nltk.corpus import stopwords
import matplotlib.pyplot as plt
from nltk.corpus import stopwords
from wordcloud import WordCloud
```

```python
# Importer Dataframe X_train
X_train=pd.read_csv('../resources/generated/X_train_cleaned.csv', index_col=0)
X_train.head()
```

```python
# Importer Dataframe Y_train
Y_train=pd.read_csv('../resources/Y_train_CVw08PX.csv', index_col=0)
Y_train.head()
```

```python
## Fusion des dataFrame:
df_train=X_train.join(Y_train)
display(df_train.head())
```

```python
dic_corresp = {
    10: 'Livres occasions',
    2280: 'Journaux, revues d\'occasion',
    50: 'accessories gaming',
    1280:'Jouets petite enfance',
    2705: 'Livres neufs',
    2522:'Fournitures scolaires',
    2582:'Accessoires cuisine & jardin',
    1560:'Accessoires maison & literie',
    1281:'Jeux de société',
    1920:'Accessoire maison',
    2403:'Livre, BD et magazines',
    1140:'Figurines',
    2583:'Equipement piscine de jardin',
    1180:'Figurines, jouets de petite taille',
    1300:'Modélisme',
    2462:'Jeux vidéo d\'occasion',
    1160:'Cartes de jeux',
    2060:'Décoration',  
    40:'jeux portatifs',
    60:'Consoles de jeux', 
    1320:'Accessoires bébé',
    1302:'Loisirs: camping, plage, jardin', 
    2220:'Accessoires animalerie', 
    2905:'Jeux vidéos aventure, combat',
    2585:'Bricolage et jardinage',
    1940:'Alimentaire',
    1301:'divertissements' }


# Fonction pour ajouter les catégories
def mapper_noms_produits(code):
    return dic_corresp.get(code, 'Produit_inconnu')

# Ajout d'une nouvelle colonne 'Catégorie'
df_train['Catégories'] = df_train['prdtypecode'].apply(mapper_noms_produits)

# Affichage du DataFrame 
display(df_train.head())
```

```python
# Définir le calque du nuage des mots
wc = WordCloud(background_color="black", max_words=100, max_font_size=50, random_state=42)
```

```python
prdtypecode_values = Y_train['prdtypecode'].unique()
 
for category in prdtypecode_values:
    current_category_data = df_train[df_train['prdtypecode']==category]
    current_text = ""
    for words in current_category_data.Texte:
        current_text += str(words)  # Convertir to str avant concatenation

    plt.figure(figsize=(15, 15))
    wc.generate(current_text)
    plt.imshow(wc)
    plt.title(f"Category ({category}) : {current_category_data.iloc[0]['Catégories']}")
    plt.axis('off')
    plt.show()
```

```python
# Sauvegarder la data cleaned
df_train.to_csv('../resources/generated/df_train_cleaned.csv')
```
